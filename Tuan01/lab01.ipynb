## Bai 6. Phep toan va nhan ma tran

### Code

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

### Nhận xét

- `a * b` là phép nhân từng phần tử.
  Hai Tensor phải có cùng shape hoặc có thể broadcast với nhau.

- `a @ b.T` là phép nhân ma trận
  Với:
  - `a` có shape `torch.Size([2, 3])`
  - `b.T` có shape `torch.Size([3, 2])`

- Khác biệt chính:
  - `a * b`: nhân từng phần tử tương ứng.
  - `a @ b.T`: thực hiện phép nhân ma trận theo quy tắc hàng x cột.

## Bai 8. Autograd co ban

### Code

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

### Tinh thu cong dao ham y=x^2 tai x = 3.6
- Dao ham y=x^2
    => y' = 2x
- Tai x = 3.6: 
- 2 * 3.6 = 7.2


### Doi ham thanh y = 3*x**2 + 2*x + 1

```python
import torch

x = torch.tensor(3.6, requires_grad=True)

y = 3 * x**2 + 2 * x + 1

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

## Bai 9. Gradient Descent tim he so da thuc

### Code

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

### Nhận xét
- Hệ số học được xấp xỉ [1.0176, 1.9998, 1.3179].
- So với hệ số thật [1, 2, 3], hai hệ số đầu khá gần, còn hệ số thứ ba vẫn lệch khá nhiều.

### Vai trò các lệnh
- optimizer.zero_grad(): xóa gradient cũ trước mỗi vòng lặp.
- mse.backward(): tính gradient của hàm mất mát MSE.
- optimizer.step(): cập nhật các hệ số w dựa trên gradient đã tính.

## Bai 10. Giai he phuong trinh bang Autograd

### Code

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

### Nhận xét
Nghiệm thu được:
- A ≈ 4.2086
- B ≈ 4.7913
- C ≈ 3.7913
- D ≈ 2.7913

Kiểm tra:
- A + B ≈ 9
- C - D ≈ 1
- A + C ≈ 8
- B - D ≈ 2

=> Nghiệm tìm được thỏa mãn bốn phương trình.

Bài toán có thể có nhiều nghiệm vì các phương trình không độc lập với nhau. Một số phương trình có thể suy ra từ các phương trình còn lại, nên hệ không xác định duy nhất một bộ giá trị A, B, C, D

```python
x = torch.tensor(2.0, requires_grad=True)

y = 2 * x**2 + 5 * x + 3

y.backward()

print("y =", y)
print("Gradient tại x = 2:", x.grad)
```

### Yeu cau 4: Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch

polynomial = np.poly1d([2, -3, 5])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(2000):
    optimizer.zero_grad()

    y_pred = x @ w

    mse = torch.mean((y - y_pred) ** 2)

    mse.backward()

    optimizer.step()

print("\nLearned coefficients:")
print(w)
```

- He so tim duoc: [2, -3, 5]
